### Listando los circuitos de carrera de la fromula 1 

```python
df_circuits = pd.read_csv("../data/circuits.csv" , index_col = 0)
df_circuits
```

### Contar Circuitos por Paises

```python
circuits_by_country_count = df_circuits.groupby("country").size()
circuits_by_country_count
```

### Contando el numero de elementos (circuitos) por grupo

```python
circuits_by_country_element = df_circuits.groupby("country").count()
circuits_by_country_element
```

### Generar estadisticas descriptivas para columnas numericas de los paises

```python
circuits_by_country_count.describe()
```

### Limpieza de Datos
