#### Serie en Pandas

Una serie Pandas es una matriz etiquetada unidimensional capaz de contener datos de cualquier tipo. Se puede crear a partir de listas de Python, matrices NumPy, diccionarios, valores escalares o datos cargados desde fuentes externas como archivos CSV, archivos Excel y bases de datos. Las etiquetas asociadas a una Serie se denominan su índice.

```python
data  = np.array(['D','M'])
s = pd.Series(data)
print("Series de Panda:")
print(s)
```

```python
#Nuevo ejemplo de trabajo 
data = {
    "Nombre" : [ "Ana" , "Juan" , "Pedro", "Maria","Luis","Carlos"],
    "Edad" : [22,25,28,23,20,23],
    "Ciudad" : ["valencia","cabudare","barcelona","barquisimeto","caracas","barcelona"],
    "Puntuacion" : [80 ,90 , 85 ,88 ,75,80]
}

df = pd.DataFrame(data)
df
```

* ##### Agregar y eliminar columnas:
puede crear nuevas columnas basadas en las existentes o eliminar columnas no deseadas del DataFrame.

* ##### grupación de datos (GroupBy):
la agrupación le permite organizar los datos en categorías y calcular valores para cada grupo, por ejemplo, sumas, recuentos o promedios.

```python
#agrupo datos por ciudad
grouped = df.groupby("Ciudad")
print(grouped.groups)


# Agregacion de Datos 
#calcular la suma de las edades y puntuacion por ciudad
agregate_data = grouped.agg({
    "Edad" : "mean" , #mean : PROMEDIO
    "Puntuacion" : "sum"
})

agregate_data
```

Agregacion de funciones Personalizadas

```python
def rango(series):
    return series.max() - series.min()

##aplicar la funcion
agregate_data_custom = grouped.agg({
    "Edad" : rango,
    "Puntuacion" : rango
  })
print(agregate_data_custom)
```

```python
df = pd.DataFrame(
    {
        "A": ["foo", "bar", "foo", "bar", "foo", "bar", "foo", "foo"],
        "B": ["one", "one", "two", "three", "two", "two", "one", "three"],
        "C": np.random.randn(8),
        "D": np.random.randn(8),
    }
)


df
```

En un DataFrame, obtenemos un objeto GroupBy llamando agroupby() . Este método devuelve una pandas.api.typing.DataFrameGroupBy instancia. Naturalmente, podríamos agrupar por columnas o, o por ambas 
 ( mas arriba se puede visualizar un pandas.api.typing.DataFrameGroupBy)

```python
grouped_one = df.groupby("A")
grouped = df.groupby("B")
grouped = df.groupby(["A","B"])

```

DataFrame groupby siempre opera a lo largo del eje 0 (filas). Para dividir por columnas, primero haga una transposición:

```python
def get_letter_type(letter) :
    if letter.lower() in 'aeiou':
        return 'vowel'
    else:
        return 'consonant'

grouped = df.T.groupby(get_letter_type)
```
