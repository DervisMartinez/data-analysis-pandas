#### Serie en Pandas

Una serie Pandas es una matriz etiquetada unidimensional capaz de contener datos de cualquier tipo. Se puede crear a partir de listas de Python, matrices NumPy, diccionarios, valores escalares o datos cargados desde fuentes externas como archivos CSV, archivos Excel y bases de datos. Las etiquetas asociadas a una Serie se denominan su índice.

```python
data  = np.array(['D','M'])
s = pd.Series(data)
print("Series de Panda:")
print(s)
```

```python
#Otro Ejemplo Seria
numbers = [3,4,5,6,7]
series = pd.Series(numbers)
series , type(series)
```

#### Pandas Data Frame

Es una estructura de datos bidimensional con ejes etiquetados (filas y columnas). Se crea cargando los conjuntos de datos del almacenamiento existente, que puede ser una base de datos SQL, un archivo CSV o un archivo Excel. Se puede crear a partir de listas, diccionarios, una lista de diccionarios, etc.


```python
df = pd.DataFrame()
print(df)
lst = ['Deris' ,'Martinez' ,'Data','Science' , 'Student']
df = pd.DataFrame(lst)
print(df)
```

#### Operaciones en Pandas

Pandas proporciona operaciones esenciales para trabajar con datos estructurados de manera eficiente. Las secciones siguientes presentan las funcionalidades más utilizadas con breves explicaciones y ejemplos sencillos

```python
# Generar diccionario
data = {
    "Nombre" : ["Ana" , "Juan" , " Pedro" , "Maria", " Luis" , "Omar" , "Eliezer"],
    "Edad" : [23,25,32,21,26,24,21],
    "Ciudad" : ["Barcelona" , "Caracas","Valencia", "Caracas","Barcelona","Miranda","Valencia"]
}
data ,type(data)
```

```python
#Generar un Data Frame apartir de un diccionario
df = pd.DataFrame(data=data)
df
```
